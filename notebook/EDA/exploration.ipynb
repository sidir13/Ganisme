## SPLITTING

```python
import os, random, shutil, pathlib
random.seed(42)

def split_folder(src, train_dst, test_dst, test_ratio=0.1):
    os.makedirs(train_dst, exist_ok=True); os.makedirs(test_dst, exist_ok=True)
    files = [f for f in os.listdir(src) if f.lower().endswith(('.jpg','.png','.jpeg'))]
    random.shuffle(files)
    k = int(len(files)*test_ratio)
    test, train = files[:k], files[k:]
    for f in train: shutil.move(os.path.join(src,f), os.path.join(train_dst,f))
    for f in test:  shutil.move(os.path.join(src,f), os.path.join(test_dst,f))

split_folder("../../data/processed/french", "../../data/processed/french/train", "../../data/processed/french/test")
split_folder("../../data/processed/dutch", "../../data/processed/dutch/train", "../../data/processed/dutch/test")
```

```python
import os, PIL.Image as Image, statistics as st
def sizes(folder):
    wh=[]; 
    for f in os.listdir(folder):
        if f.lower().endswith(('.jpg','.png','.jpeg')):
            w,h = Image.open(os.path.join(folder,f)).size; wh.append((w,h))
    return wh

for d in ["train","test"]:
    wh = sizes(f"../../data/processed/paint_portrait/{d}")
    print(d, len(wh), "images — médiane:", (st.median([w for w,h in wh]), st.median([h for w,h in wh])))
```
