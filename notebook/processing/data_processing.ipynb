```python

output_dir = "../data/paint_portrait"
os.makedirs(output_dir, exist_ok=True) 

def download_wga_image(page_url, output_dir, sleep_time=1.5):
    img_path = page_url.replace("/html/", "/detail/")
    img_url = img_path.replace(".html", ".jpg")


    filename = os.path.join(output_dir, os.path.basename(img_url))


    if os.path.exists(filename):
        print(f"L'image existe déjà : {filename}, téléchargement ignoré.")
        return


    response = requests.get(img_url)
    if response.status_code == 200:
        with open(filename, "wb") as f:
            f.write(response.content)
        print(f"Image téléchargée : {filename}")
        time.sleep(sleep_time)  
    else:
        print("Impossible de télécharger l'image :", response.status_code, img_url)
```

```python
for url in urls:
    download_wga_image(url.strip(), output_dir)
```

```python
import pandas as pd

# Si ton fichier est en CSV :
data_file = "../../data/art_catalog.xlsx"
df = pd.read_excel(data_file)

print(df.columns)  # pour vérifier que "TYPE" et "URL" existent bien
```

```python
# On garde les lignes où :
#   - FORM contient "painting"
#   - TYPE contient "portrait"
df_portraits = df[
    df["FORM"].str.lower().str.contains("painting", na=False)
    & df["TYPE"].str.lower().str.contains("portrait", na=False)
].copy()

print(f"Nombre total de lignes : {len(df)}")
print(f"Nombre de portraits : {len(df_portraits)}")
```

```python
def page_url_to_filename(page_url: str) -> str:
    img_url = page_url.replace("/html/", "/detail/").replace(".html", ".jpg")
    return img_url.split("/")[-1]

df_portraits["fname"] = df_portraits["URL"].apply(page_url_to_filename)
```

```python
import os, shutil

src_dir = "../../data/processed"
dst_dir = "../../data/paint_portrait_only"
os.makedirs(dst_dir, exist_ok=True)

copied = 0
missing = []

for fname in df_portraits["fname"].unique():
    src = os.path.join(src_dir, fname)
    dst = os.path.join(dst_dir, fname)
    if os.path.exists(src):
        shutil.copy2(src, dst)
        copied += 1
    else:
        missing.append(fname)

print(f"✅ {copied} portraits copiés dans {dst_dir}")
print(f"⚠️ {len(missing)} fichiers manquants (peut-être non téléchargés)")
```
