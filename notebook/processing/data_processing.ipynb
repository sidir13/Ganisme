```python
import pandas as pd

# Si ton fichier est en CSV :
data_file = "../../data/art_catalog.xlsx"
df = pd.read_excel(data_file)

print(df.columns)  # pour vérifier que "TYPE" et "URL" existent bien
```

```python
# On garde les lignes où :
#   - FORM contient "painting"
#   - TYPE contient "portrait"
df_portraits = df[
    df["FORM"].str.lower().str.contains("painting", na=False)
    & df["TYPE"].str.lower().str.contains("landscap", na=False)
].copy()

print(f"Nombre total de lignes : {len(df)}")
print(f"Nombre de portraits : {len(df_portraits)}")
```

```python
def page_url_to_filename(page_url: str) -> str:
    img_url = page_url.replace("/html/", "/detail/").replace(".html", ".jpg")
    return img_url.split("/")[-1]

df_portraits["fname"] = df_portraits["URL"].apply(page_url_to_filename)
```

```python
import os, shutil

src_dir = "../../data/raw/all"
dst_dir = "../../data/processed/landscapes"
os.makedirs(dst_dir, exist_ok=True)

copied = 0
missing = []

for fname in df_portraits["fname"].unique():
    src = os.path.join(src_dir, fname)
    dst = os.path.join(dst_dir, fname)
    if os.path.exists(src):
        shutil.copy2(src, dst)
        copied += 1
    else:
        missing.append(fname)

print(f"✅ {copied} portraits copiés dans {dst_dir}")
print(f"⚠️ {len(missing)} fichiers manquants (peut-être non téléchargés)")
```
